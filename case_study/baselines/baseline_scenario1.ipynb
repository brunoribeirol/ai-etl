```python
import time
from pathlib import Path

import pandas as pd

start = time.time()
```

### 1. Extração

```python
df = pd.read_csv("../data/sales.csv")
print(f"Loaded: {df.shape}")
df.head(3)
```

### 2. Transformação

```python
# Rename columns
df = df.rename(columns={"dt": "date", "amt": "amount"})

# Filter active only
df = df[df["status"] == "active"]

# Remove duplicates
df = df.drop_duplicates().reset_index(drop=True)

print(f"After transform: {df.shape}")
df.head(3)
```

### 3. Qualidade (manual)

```python
null_report = df.isnull().mean().round(4)
dup_count = df.duplicated().sum()

print("Null ratios:")
print(null_report[null_report > 0])
print(f"\nDuplicates remaining: {dup_count}")
```

### 4. Carga

```python
out_path = Path("../results/scenario1/output_baseline.csv")
out_path.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(out_path, index=False)

elapsed = time.time() - start
print(f"Saved {len(df)} rows → {out_path}")
print(f"Pipeline execution time: {elapsed:.2f}s")
```
